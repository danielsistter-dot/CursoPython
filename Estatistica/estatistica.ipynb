```python
dict_metrics = {"idade":[15,18,25,26,40,40,58,60],
                "altura":[160,165,149,172,174,174,178,168]}
```

```python
type(dict_metrics)
```

```python
df=pd.DataFrame(dict_metrics)
```

```python
type(df)
```

```python
df
```

```python
print(f"Media de idade: {df['idade'].mean()}")
print(f'Media de altura: {df["altura"].mean()}')
```

```python
print(f"Mediana de idade: {df['idade'].median()}")
print(f'Mediana de altura: {df["altura"].median()}')
```

```python
print(f"Media de idade: {df['idade'].mean()}")
print(f'Media de altura: {df["altura"].mean()}')
```

```python
print(f"Moda de idade: {df['idade'].mode()[0]}")
print(f'Moda de altura: {df["altura"].mode()}')
```

```python
print(f"Variancia de idade: {df['idade'].var()}")
print(f'Variencia de altura: {df["altura"].var()}')
```

```python
print(f"Desvio padrao de idade: {df['idade'].std()}")
print(f'Desvio padrao de altura: {df["altura"].std()}')
```

```python
df["idade"].std()  /  df["idade"].mean()  * 100
```

```python
df["altura"].std()  /  df["altura"].mean()  * 100
```
