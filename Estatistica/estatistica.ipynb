```python
print(f"Media de idade: {df['idade'].mean()}")
print(f'Media de altura: {df["altura"].mean()}')
```

```python
dic_vendas ={"categorias":["eletrodomesticos","games","livros","casa"],
            "valor":[2000,50000,10000,200000],
            "QUANTIDADE":[200,500,300,1500]
}
```

```python
df_vendas= pd.DataFrame(dic_vendas)
```

```python
df_vendas
```

```python
df_vendas.dtypes
```

```python
df_vendas.plot.bar(x="categorias",y="valor")
```

```python
df_vendas.sort_values("valor", ascending=True).plot.bar(x="categorias",y="valor")
```

```python
df_vendas.sort_values("QUANTIDADE", ascending=True).plot.barh(x="categorias",y="QUANTIDADE")
```

```python
dic_faturamento={"data":["2020-01-01","2021-01-01","2022-01-01","2023-01-01","2024-01-01"],
                 "valor":["100000","200000","150000","300000","350000"]}
```

```python
df_faturamento=pd.DataFrame(dic_faturamento)
```

```python
df_faturamento.dtypes
```

```python
df_faturamento["data"]=pd.to_datetime(df_faturamento["data"])
```

```python
df_faturamento["valor"]=pd.to_numeric(df_faturamento["valor"])
```

```python
df_faturamento.dtypes
```

```python
df_faturamento.plot.line(x="data",y="valor")
```
